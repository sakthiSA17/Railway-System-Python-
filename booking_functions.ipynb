```python
def bill():
    fr = {
        "a1": ("apple", 150),
        "a2": ("banana", 60),
        "a3": ("orange", 120),
        "a4": ("grapes", 95),
        "a5": ("guava", 45),
    }
    total = 0
    while True:
        x = input("Fruits code (a1-a5): ").strip().lower()
        r = float(input("Enter the quantity: "))

        if x in fr:
            name, price = fr[x]
        else:
            name, price = "other", 90

        amount = r * price
        print(f"{name} - {r} x {price} = {amount}")
        total += amount

        more = input("Add another fruit? (yes/no): ").strip().lower()
        if more != "yes":
            break
    return total

def discount(o):
    if o >= 5000:
        f = o * 20 / 100
        t = o - f
        print("Your total amount for purchase:", o)
        print("Your discount of 20%:", f)
        print("Your amount after discount:", t)
    else:
        print("You are not eligible for a discount")
        print("Your total amount for purchase:", o)
    print("Thank you for your purchase")

discount(bill())
```
